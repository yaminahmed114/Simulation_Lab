```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python
#Task-1. Hypotheses
print("Task 1: Hypotheses")
print("H0: mu1 - mu2 = 0 (Population means are equal)")
print("H1: mu1 - mu2 != 0 (Population means are not equal)\n")
```

```python
#Task-2. Sample Means, Variances, and Mean Difference
data1 = np.array([65, 78, 56, 68, 72, 70, 62, 75, 80, 67])
data2 = np.array([70, 82, 60, 75, 78, 73, 68, 80, 85, 72])

mean1 = np.mean(data1)
mean2 = np.mean(data2)
var1 = np.var(data1, ddof=1)
var2 = np.var(data2, ddof=1)
mean_diff = mean1 - mean2

print("--- Task 2: Means & Variances ---")
print(f"Sample 1 Mean: {mean1:.4f}")
print(f"Sample 2 Mean: {mean2:.4f}")
print(f"Sample 1 Variance (ddof=1): {var1:.4f}")
print(f"Sample 2 Variance (ddof=1): {var2:.4f}")
print(f"Mean Difference (Data1 - Data2): {mean_diff:.4f}\n")
```

```python
#Task-3. t-Statistic, Degrees of Freedom, and Two-Tailed p-Value
n1, n2 = len(data1), len(data2)
df = n1 + n2 - 2

# Pooled variance and standard error calculations
sp_sq = ((n1 - 1) * var1 + (n2 - 1) * var2) / df
std_err = np.sqrt(sp_sq * (1/n1 + 1/n2))

t_stat, p_val = stats.ttest_ind(data1, data2, equal_var=True)

print("--- Task 3: Test Statistics ---")
print(f"Degrees of Freedom (df): {df}")
print(f"t-Statistic: {t_stat:.4f}")
print(f"Two-Tailed p-Value: {p_val:.6f}\n")
```

```python
#Task-4. 95% Confidence Interval for Mean Difference
ci = stats.t.interval(0.95, df, loc=mean_diff, scale=std_err)

print("--- Task 4: 95% Confidence Interval ---")
print(f"95% CI: [{ci[0]:.4f}, {ci[1]:.4f}]\n")
```

Task 5: Decision & Interpretation
Decision: Reject H0 (p-value 0.000027 < alpha 0.05)
Interpretation: There is strong evidence that the population means differ significantly. Data 2 is significantly higher than Data 1.
